```python
marks = int(input("Enter marks your marks: "))

if marks >= 80:
    print("A")
elif marks >= 70:
    print("B")
elif marks >= 60:
    print("C")
elif marks >= 50:
    print("D")
else:
    print("F")
print("this is grade checking system")
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
c=set1.union(set2)
d=set1.intersection(set2)
e=set1.difference(set2)
f=set1.symmetric_difference(set2)
print(c,d,e,"\n",f)
```
